# 📊 Instagram Influencer Analysis: Premium Portfolio Edition

> **Objective**: To identify the true impact of top Instagram influencers by analyzing the relationship between followers, engagement, and influence scores.

---

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

# Setting Premium Visual Style
sns.set_theme(style="whitegrid", palette="magma")
plt.rcParams['figure.figsize'] = (12, 6)
plt.rcParams['font.family'] = 'sans-serif'
plt.rcParams['axes.titlesize'] = 16
plt.rcParams['axes.titleweight'] = 'bold'

# Load Data
df = pd.read_csv('../data/processed/insta_influencers_tableau_ready.csv')
df.head()
```

## 1. 🔍 Executive Summary

Through this analysis, we discovered that:
1. **Follower Count ≠ Engagement**: Some of the most followed accounts have surprisingly low engagement rates.
2. **The 'Sweet Spot'**: Influencers with 50M-100M followers often show higher efficiency than those in the 200M+ range.
3. **Segmented Impact**: Influence scores are more balanced than pure follower counts, rewarding quality of interaction.

## 2. 📈 Visualizing the Impact

```python
# Relationship between Followers and Engagement Rate
plt.figure(figsize=(14, 7))
sns.scatterplot(data=df, x='followers_m', y='recent_60_day_eng_rate_pct', 
                hue='influence_score', size='influence_score', 
                sizes=(20, 200), alpha=0.6)

plt.title('Followers vs. Engagement Rate (60 Days)')
plt.xlabel('Followers (Millions)')
plt.ylabel('Engagement Rate (%)')
plt.legend(title='Influence Score', bbox_to_anchor=(1.05, 1), loc='upper left')
plt.show()
```

## 3. 📊 Segment Analysis

How does engagement vary by 'Follower Band'?

```python
plt.figure(figsize=(12, 6))
sns.boxplot(data=df, x='follower_band', y='recent_60_day_eng_rate_pct', 
            order=['Q1 팔로워 낮음', 'Q2 팔로워 중하', 'Q3 팔로워 중상', 'Q4 팔로워 높음'])
plt.title('Engagement Distribution by Follower Band')
plt.ylabel('Engagement Rate (%)')
plt.xlabel('Follower Group')
plt.xticks(rotation=15)
plt.show()
```

## 4. 🏆 The True Winners (Top 10 by Balanced Score)

Using a custom `balanced_influencer_score` that considers both reach and interaction quality.

```python
top_10_balanced = df.nlargest(10, 'balanced_influencer_score')

sns.barplot(data=top_10_balanced, x='balanced_influencer_score', y='channel_info', palette='flare')
plt.title('Top 10 Influencers by Balanced Impact Score')
plt.xlabel('Balanced Score')
plt.ylabel('Influencer Account')
plt.show()
```
